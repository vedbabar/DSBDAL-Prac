```python
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
with open("file.txt","r") as  f:
    doc1 = f.read()
```

```python
print(doc1)
```

```python
tokens= word_tokenize(doc1)
print(tokens)
```

```python
pos_tags = pos_tag(tokens)
print(pos_tags)
```

```python
stop_words = set(stopwords.words('english'))
filtered_tokens=[]
for word in tokens:
    if word.lower() not in stop_words:
        filtered_tokens.append(word)
print(filtered_tokens)
```

```python
stemmer = PorterStemmer()
stemmed_words=[]
for word in filtered_tokens:
    stemmed_words.append(stemmer.stem(word))
print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words=[]
for word in filtered_tokens:
    lemmatized_words.append(lemmatizer.lemmatize(word))
print(lemmatized_words)
```

```python
from collections import Counter
word_counts = Counter(filtered_tokens)
total_words = len(filtered_tokens)
tf = {} 

# 2. Go through every word and its count one by one
for word, count in word_counts.items():
    # 3. Calculate the frequency: count divided by total
    frequency = count / total_words
    # 4. Put the result into our new dictionary
    tf[word] = frequency

print("Term Frequency:", tf)
```

```python
vectorizer = TfidfVectorizer()
tfid_matrix = vectorizer.fit_transform([doc1])
feature_names = vectorizer.get_feature_names_out()
dense = tfid_matrix.todense()
denselist = dense.tolist()
df = pd.DataFrame(denselist)
df.head()
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Our Corpus (List of documents)
doc1 = "spark is fast"
doc2 = "spark is powerful"
corpus = [doc1, doc2] # You MUST pass multiple docs here

# 2. Initialize and Fit the Vectorizer
vectorizer = TfidfVectorizer()
tfid_matrix = vectorizer.fit_transform(corpus)

# 3. Extract the feature names (the vocabulary)
feature_names = vectorizer.get_feature_names_out()

# 4. Convert to a readable Pandas DataFrame
dense = tfid_matrix.todense()
df = pd.DataFrame(dense, columns=feature_names, index=["Doc 1", "Doc 2"])

print(df)
```
