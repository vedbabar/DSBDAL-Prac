```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix , accuracy_score , classification_report
```

```python
url = r"C:\Users\vedba\Documents\DSBDAL_prac\iris.csv"
df = pd.read_csv(url)
df.head(10)
```

```python
x=df.drop(['variety'], axis=1)
y=df['variety']

x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
nb_model = GaussianNB()
nb_model.fit(x_train,y_train)

y_pred = nb_model.predict(x_test)
print(y_pred[:10])
```

```python
cm = confusion_matrix(y_test,y_pred)
print(cm)
```

```python
# 2. Extract TP, FP, FN, TN using matrix operations
# True Positives: The diagonal elements
TP = np.diag(cm)

# False Positives: Column sum minus the diagonal (TP)
FP = np.sum(cm, axis=0) - TP

# False Negatives: Row sum minus the diagonal (TP)
FN = np.sum(cm, axis=1) - TP

# True Negatives: Total matrix sum minus (TP + FP + FN)
TN = np.sum(cm) - (FP + FN + TP)

# 3. Apply the Mathematical Formulas
# Overall Accuracy: Total correct predictions over total overall predictions
total_predictions = np.sum(cm)
overall_accuracy = np.sum(TP) / total_predictions
overall_error_rate = 1 - overall_accuracy

# Precision and Recall per class 
# (Adding a tiny value 1e-6 to avoid Division by Zero if a class has 0 predictions)
precision = TP / (TP + FP + 1e-6)
recall = TP / (TP + FN + 1e-6)

# 4. Print the final computed values
print("=== Overall Metrics ===")
print(f"Overall Accuracy: {overall_accuracy * 100:.2f}%")
print(f"Overall Error Rate: {overall_error_rate * 100:.2f}%\n")

print("=== Class-Wise Metrics (Setosa, Versicolor, Virginica) ===")
print(f"True Positives (TP): {TP}")
print(f"False Positives (FP): {FP}")
print(f"False Negatives (FN): {FN}")
print(f"True Negatives (TN): {TN}\n")

# Formatting precision and recall arrays for clean printing
precision_pct = [f"{val * 100:.2f}%" for val in precision]
recall_pct = [f"{val * 100:.2f}%" for val in recall]

print(f"Precision per class: {precision_pct}")
print(f"Recall per class:    {recall_pct}")
```
